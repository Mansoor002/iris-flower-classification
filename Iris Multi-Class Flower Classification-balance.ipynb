```python
#Creating dummy variables and stores the result in the new DataFrame "Dataset"
dataset = pd.get_dummies(df, columns = ['Id'])
```

```python
#importing necessary modules for cross-validation and data preprocessing, then standardizes the selected columns using StandardScaler from scikit-learn.
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
standardScaler = StandardScaler()
columns_to_scale = ['PetalLengthCm', 'PetalWidthCm']
dataset[columns_to_scale] = standardScaler.fit_transform(dataset[columns_to_scale])
```

```python
#Shows first 5 columns of "dataset"
dataset.head()
```

```python
#Defines the features "X" and the target variable "y" by separating the 'Species' column from the rest of the DataFrame "dataset".
y = dataset['Species']
X = dataset.drop(['Species'], axis = 1)
```

# RANDOM FOREST CLASSIFIER

```python
#importing the RandomForestClassifier class from the sklearn.ensemble module, which is used for building random forest models.
from sklearn.ensemble import RandomForestClassifier
```

```python
#creating a RandomForestClassifier with 10 estimators and evaluates its performance using 10-fold cross-validation.
randomforest_classifier= RandomForestClassifier(n_estimators=10)

score=cross_val_score(randomforest_classifier,X,y,cv=10)
```

```python
#calculating the mean of the scores obtained from cross-validation.
score.mean()
```

# DECISION TREE CLASSIFIER

```python
#importing the DecisionTreeClassifier class from the sklearn.tree module, which is used for building decision tree models.

from sklearn.tree import DecisionTreeClassifier
```

```python
#initializing a DecisionTreeClassifier and evaluates its performance using 15-fold cross-validation, storing the scores in the variable "score".

decisiontree_classifier = DecisionTreeClassifier()

score = cross_val_score(decisiontree_classifier,X,y,cv=15)
```

```python
#calculating the mean of the scores obtained from cross-validation.

score.mean()
```

# CONFUSION MATRIX

```python
from sklearn import metrics

actual = np.random.binomial(1,.9,size = 1000)
predicted = np.random.binomial(1,.9,size = 1000)

confusion_matrix = metrics.confusion_matrix(actual, predicted)

cm_display = metrics.ConfusionMatrixDisplay(confusion_matrix = confusion_matrix, display_labels = [0, 1])

cm_display.plot()
plt.show()
```
